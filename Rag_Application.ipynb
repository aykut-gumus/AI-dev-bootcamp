```python
def sor(kullanici, sistem=None):
  if sistem:
    prompt = f"SİSTEM TALİMATI : \n{sistem}\n\nKULLANICI MESAJI:\n{kullanici}"
  else:
    prompt = kullanici
  return ai.generate_text(
      prompt,
      model_name = model
  )
```

```python
import json
from typing import Literal
from pydantic import BaseModel
```

```python
class Talep(BaseModel):
  departman: Literal["Fatura", "Teknik", "Kargo", "İade"]
  aciliyet: Literal["Düşük", "Orta", "Yüksek"]
```

```python
def json_coz(metin, sema):
  return sema.model_validate_json(metin[metin.find("{"):metin.rfind("}")+1])
```

```python
istek = f"""Müşteri mesajını analiz et. YALNIZCA şu şemaya uyan tek bir JSON yaz:
{json.dumps(Talep.model_json_schema(), ensure_ascii=False)}
Mesaj: Faturamda tanımadığım 300 TL'lik bir kalem var, acilen açıklayın!"""
```

```python
import csv
from pypdf import PdfReader
from langchain_core.documents import Document
```

```python
def pdf_yukle(yol, **ek_bilgi):
  return [Document(page_content=sayfa.extract_text(),
                   metadata={"kaynak": yol, "sayfa": no, **ek_bilgi})
            for no, sayfa in enumerate(PdfReader(yol).pages, start=1)]
```

```python
def metin_yukle(yol, **ek_bilgi):
  with open(yol, encoding = "utf-8") as f:
    return [Document(page_content="\n".join(f"{k}: {v}" for k, v in satir.items()),
                     metadata={"kaynak": yol, "satir": no, **ek_bilgi})
              for no, satir in enumerate(csv.DictReader(f), start=1)]
```

```python
def csv_yukle(yol, **ek_bilgi):
  with open(yol, encoding="utf-8") as f:                                  # Dosyayı aç.
        return [Document(page_content="\n".join(f"{k}: {v}" for k, v in satir.items()),   # satır → metin,
                         metadata={"kaynak": yol, "satir": no, **ek_bilgi})              # hangi satır.
                for no, satir in enumerate(csv.DictReader(f), start=1)]
```

```python
import re
from langchain_text_splitters import (
    CharacterTextSplitter,
    RecursiveCharacterTextSplitter,
    MarkdownHeaderTextSplitter
)
```

```python
sabit = CharacterTextSplitter(separator="", chunk_size=300, chunk_overlap=50).split_documents(el_2026)
```

```python
yinelemeli = RecursiveCharacterTextSplitter(chunk_size=400, chunk_overlap=60).split_documents(el_2026)
```

```python
def bolumlere_ayir(dokumanlar):
  parcalar = []
  for d in dokumanlar:
    for bolum in re.split(r"\n(?=\d+\.[A-ZÇĞİÖŞÜ])", d.page_content):
      bolum = bolum.strip()
      if not re.match(r"\d+\. ", bolum):
        continue
      parcalar.append(Document(page_content=bolum,
                               metadata={**d.metadata, "bolum":bolum.split("\n")[0]}))
    return parcalar
```

```python
bolumler_2026 = bolumlere_ayir(el_2026)
```

```python
import numpy as np
```

```python
def ilk_rag(soru, parcalar, k=3):
  metinler = [p.page_content for p in parcalar]
  V = mpnet.encode(metinler, normalize_embeddings=True)
  q = mpnet.encode(soru, normalize_embeddings=True)
  benzerlikler = V @ q

  en_iyi = np.argsort(-benzerlikler)[:k]
  for i in en_iyi:
    skor = float(benzerlikler[i])
    # 'bolum' yoksa 'sayfa' bilgisini göster, o da yoksa 'Bilinmiyor' yaz
    kaynak_bilgisi = parcalar[i].metadata.get('bolum', f"Sayfa {parcalar[i].metadata.get('sayfa', '?')}")
    print(f"{skor:.2f} | {kaynak_bilgisi}")

  baglam = "\n\n".join(metinler[i] for i in en_iyi)
  return sor(f"Bağlam:\n{baglam}\n\nSoru:{soru}",
             sistem="Yalnızca bağlama göre cevapla; bağlamda yoksa 'Dokümanda yok' de."
             )
```

```python
for soru in ["Tatile kaç gün çıkabilirim?", "Bilgisayarım bozuldu, ne yapmalıyım?"]:
  print("Soru:", soru)
  print("Cevap:", ilk_rag(soru, yinelemeli).strip())
  print("-" * 40)
```
